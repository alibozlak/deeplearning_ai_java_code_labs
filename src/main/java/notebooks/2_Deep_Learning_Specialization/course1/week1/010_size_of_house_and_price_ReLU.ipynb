### ReLU function with Size of House feature and Price sample

We told ***Single Variable Linear Regression*** in 1_Machine_Learning_Specialization/course1/week1 with ***Size of House feature*** and ***Price target***. But ***linear regression prediction can be negatif value***. We solution with ***ReLU*** function the problem : <br>

$ z = f(feature) = weight \cdot feature + bias $<br>

$ ReLU(z) = max(0,z) \ge 0 $<br>

$ Model = ReLU \circ f $ <br>

$ Model(feature) = ReLU( \ f(feature) \ ) \ge 0 $<br>

```python
import deeplearning_ai.machine_learning_specialization.course1.week1.*;
import deeplearning_ai.deep_learning_specialization.course1.week1.*;

import java.util.function.Function;
import static java.lang.System.out;

import org.nd4j.linalg.api.ndarray.INDArray;
import org.nd4j.linalg.factory.Nd4j;
```

Let's $ \ f(sizeOfHouse) = 100 \cdot sizeOfHouse - 150 \ $ .

```python
float weight = 100.f;
float bias = -150.f;
SingleVariablePredictFunction singleVariablePredictionFunc = SingleVariablePredictFunctionImpl.create(weight,bias);


float sizeOfHouse1 = 1.2f; //Unit : 1000 square feet
INDArray houseSizes = Nd4j.create(new float[]{sizeOfHouse1});

INDArray predictions = singleVariablePredictionFunc.apply(houseSizes);
float predictionPriceForHouse1 = predictions.getFloat(0);

out.println("predictionPriceForHouse1 = " + predictionPriceForHouse1);  //predictionPriceForHouse1 = -29.999992  // < 0

// With ReLU : 
Function<INDArray,INDArray> relu = ReLUForSingleVariable.create();
INDArray predictionsWithReLU = relu.apply(predictions);
float predictionHouse1WithRelu = predictionsWithReLU.getFloat(0);

out.println("predictionHouse1WithRelu = " + predictionHouse1WithRelu);  //predictionHouse1WithRelu = 0.0
```
