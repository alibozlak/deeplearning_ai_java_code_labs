### Multivariable (Single Value) Functions

So far, we studied single variable (has single feature) functions. (***sizeOfHouse -> f -> price***.) <br> 
Real life has multivariable, like ***(sizeOfHouse, bedroomCount, downtownDistance, ..) -> f -> price***.<br>

> Note : I will use Andrew Ng's notation : <br><br>
<img src="./images/020_multiple_feature.png" alt="Multiple Feature Sample" width="1024"> <br>

- $\ \vec{x}^{(i)} = \begin{bmatrix} x_1^{(i)} \\ x_2^{(i)} \\ x_3^{(i)} \\ x_4^{(i)} \end{bmatrix}\  is\ i.\ sample\ feature\ array,\ \vec{w} = \begin{bmatrix} w_1 \\ w_2 \\ w_3 \\ w_4 \end{bmatrix}\ is\ f\ model's\ weight\ array\ (weight\ parameters),\ $<br>

$ b\ is\  f\ model's\ bias\ parameter,\  $<br>

$ y^{(i)}\ is\ i.\ sample\ target,\  $<br>

$ \hat{y}^{(i)}\ is\ model's\ prediction\ for\ i.\ sample.\ $<br>

- $\ Our\ model:\ f(\vec{x}^{(i)}) = w_1x_1{(i)} + w_2x_2^{(i)} + w_3x_3^{(i)} + w_4x_4^{(i)} + b\ =\ \vec{w}\cdot\vec{x}^{(i)} + b $<br>

$ \hat{y}^{(i)}\ = f(\vec{x}^{(i)}) $<br><br>

$ Our\ Loss\ Function\ for\ i.\ sample:\  L^{(i)}(w_1,w_2,w_3,w_4,b) = L^{(i)}(\vec{w},b) = (\hat{y}^{(i)} - y^{(i)})^2 $<br>

> $ \hat{y}^{(i)}\ $ depends on model's parameters $\ (\vec{w}\ and\ b).\qquad y^{(i)}\ $ is constant. 


#### Compute Derrivatives : 

$ \frac{\partial L^{(i)}(\vec{w},b)}{\partial w_j} = \frac{\partial L^{(i)}(\vec{w},b)}{\partial \hat{y}^{(i)}} \cdot \frac{\partial \hat{y}^{(i)}}{\partial w_j}\ $ (Same logic of : [week1 notebook 031](../../week1/031_gradient_descent_lab.ipynb))(Chain Rule) <br>

$ = (2\cdot (\hat{y}^{(i)} - y^{(i)}))\cdot x_j^{(i)} \qquad  (\frac{\partial \hat{y}^{(i)}}{\partial w_j}\ = x_j^{(i)})(Other\ all\ w\ is\ constant.\ Only\ w_j\ is\ variable\ here.)\ $<br>

$ = 2\cdot (f(\vec{x}^{(i)}) - y^{(i)})\cdot x_j^{(i)} $<br>

$$ \frac{\partial L^{(i)}(\vec{w},b)}{\partial w_j} = 2\cdot (f(\vec{x}^{(i)}) - y^{(i)})\cdot x_j^{(i)} $$
<br><br>

$ \frac{\partial L^{(i)}(\vec{w},b)}{\partial b} = \frac{\partial L^{(i)}(\vec{w},b)}{\partial \hat{y}^{(i)}} \cdot \frac{\partial \hat{y}^{(i)}}{\partial b}\ $ <br>

$ = (2\cdot (\hat{y}^{(i)} - y^{(i)})) \cdot 1 \qquad  (\frac{\partial \hat{y}^{(i)}}{\partial b}\ = 1)(All\ w\ is\ constant\ here.)\ $<br>

$ = 2\cdot (f(\vec{x}^{(i)}) - y^{(i)})$<br>

$$ \frac{\partial L^{(i)}(\vec{w},b)}{\partial w_j} = 2\cdot (f(\vec{x}^{(i)}) - y^{(i)}) $$
<br>

$ J(\vec{w},b) = \frac{1}{2m}\cdot \sum_{i = 1}^{m} L^{(i)}(\vec{w}, b) $<br>

$\frac{\partial J(\vec{w},b)}{\partial w_j} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} \frac{\partial L^{(i)}(\vec{w},b)}{\partial w_j} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} 2\cdot (f(\vec{x}^{(i)}) - y^{(i)})\cdot x_j^{(i)} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(\vec{x}^{(i)}) - y^{(i)}) \cdot x_j^{(i)} $ <br> 

$$
\frac{\partial J(\vec{w},b)}{\partial w_j} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(\vec{x}^{(i)}) - y^{(i)}) \cdot x_j^{(i)}
$$
<br>

$\frac{\partial J(\vec{w},b)}{\partial b} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} \frac{\partial L^{(i)}(\vec{w},b)}{\partial b} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} 2\cdot (f(\vec{x}^{(i)}) - y^{(i)}) = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(\vec{x}^{(i)}) - y^{(i)}) $ <br>

$$
\frac{\partial J(\vec{w},b)}{\partial b} = \frac{1}{m}\cdot \sum_{i = 1}^{m} f(\vec{x}^{(i)}) - y^{(i)}
$$
