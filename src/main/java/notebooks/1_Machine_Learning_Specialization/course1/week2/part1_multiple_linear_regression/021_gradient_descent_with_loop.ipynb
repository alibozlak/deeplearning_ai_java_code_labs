### Gradient Descent With Loop 

```python
import static java.lang.System.out;

import deeplearning_ai.machine_learning_specialization.course1.week2.loop_programming.*;

import java.util.function.Function;
```

```python
// Our Model : 
Double[] weights = {0.1, 0.2, 0.3, 0.4};    //Arbitrary initialization
Double bias = 0.5;                          //Arbitrary
MultivariableFunction ourModel = OurModel.createModel(weights, bias);

// our dataset features and targets (prices) : 
Double[][] features = {
    // {sizeOfHouse, bedroomCount, floorCount, ageOfHouse}
    {2104., 5., 1., 45.},    // <--- First house
    {1416., 3., 2., 40.},
    {1534., 3., 2., 30.},
    {852.,  2., 1., 36.}
};
Double[] targets = {460., 232., 315., 178.};    // <--- prices 

// Our Cost Function :
double J(Double[] weightVector, Double bias) {
    int m = targets.length;     // <--- Size of dataset
    double cost = 0;
    for (int i = 0; i < m; i++) 
        cost += Math.pow(ourModel.apply(features[i]) - targets[i], 2);

    return (double)cost / (2*m);
}
```

We calculated derivatives ([week_2 notebook 020](./020_Derivatives.ipynb)) : <br>

$$
\frac{\partial J(\vec{w},b)}{\partial w_j} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(\vec{x}^{(i)}) - y^{(i)}) \cdot x_j^{(i)}
$$

$$
\frac{\partial J(\vec{w},b)}{\partial b} = \frac{1}{m}\cdot \sum_{i = 1}^{m} f(\vec{x}^{(i)}) - y^{(i)}
$$
<br>

```python
// If jThWeightIndex < 0, Calculates dJ/db. Otherwise dJ/dw_j. 
Function<Integer,Double> calculateDerivative = jThWeightIndex -> {
    int m = targets.length;     // <--- Size of house samples for training. 
    Double derivative = 0.0; 
    for (int i = 0; i < m; i++)
        if (jThWeightIndex > -1)
            derivative += (ourModel.apply(features[i]) - targets[i]) * features[i][jThWeightIndex];
        else 
            derivative += (ourModel.apply(features[i]) - targets[i]);

    return derivative / m;
};
```

```python
void applyGradientDescentWithLoop(int loopSize, double learningRate) {
    out.println("Initial J = " + J(weights, bias));
    
    int n = weights.length;     // n = feature size
    for (int loop = 0; loop < loopSize; loop++){
        Double[] newWeights = new Double[n];
        // *******************************************************************
        //  We will vectorize this step : (A step will not wait another step)  
        // ******************************************************************* 
        for (int j = 0; j < n; j++)
            newWeights[j] = ourModel.getWeights()[j] - learningRate * calculateDerivative.apply(j);
        
        Double new_b = ourModel.getBias() - learningRate * calculateDerivative.apply(-1);  
        
        
        // ***************************************************************
        // ****************  Simultanous Update :  *********************** 
        // *************************************************************** 
        ourModel.updateWeights(newWeights);
        ourModel.updateBias(new_b);
        
        out.println("Loop : " + (loop + 1) +", J = " + J(ourModel.getWeights(), ourModel.getBias()));
    }
}
```

```python
int loopSize = 4;
double learningRate = 5.0e-7;   
applyGradientDescentWithLoop(loopSize, learningRate);

// Logs : 
// Initial   J = 10740.947499999998
// Loop : 1, J = 949.3692856951917
// Loop : 2, J = 597.527796933896
// Loop : 3, J = 584.8740638725292
// Loop : 4, J = 584.407995158359
```
