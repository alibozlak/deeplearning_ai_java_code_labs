#### Vectorization with Java (ND4J)

```python
// ND4J is linear algebra library like NumPy

import org.nd4j.linalg.api.ndarray.INDArray;
import org.nd4j.linalg.factory.Nd4j;

import java.util.Random;
import static java.lang.System.out;
```

```python
Random random = new Random();
double[] randomNumbers =  random.doubles(1_000_000, -1000, 1000).toArray();

Runnable dotProductWithLoopProcess = () -> {
    double result = 0;
    int n = randomNumbers.length;
    for (int i = 0; i < n; i++)
        result += randomNumbers[i] * randomNumbers[i];
    // out.println("Loop result = " + result);
};
```

```python
double measureProccesTime(Runnable dotProductProccess){
    long start = System.nanoTime();
    dotProductProccess.run();
    long end = System.nanoTime();
    return (end-start) / 1_000_000.0;
}
```

```python
INDArray randomNDArray = Nd4j.create(randomNumbers);

Runnable dotProductWithVectorization = () -> {
    double result = Nd4j.getBlasWrapper().dot(randomNDArray, randomNDArray);
    // out.println("Vectorization result = " + result);
};
```

```python
double loopMs = measureProccesTime(dotProductWithLoopProcess);
double vectorizationMs = measureProccesTime(dotProductWithVectorization);

out.println("Loop ms          : " + loopMs);
out.println("Vectorization ms : " + vectorizationMs);

// Loop result          = 3.334926329853E11
// Vectorization result = 3.334926329853E11

// First result :
// Loop ms          : 2.7788
// Vectorization ms : 0.8058

// Later results (about) :   
// Loop ms          : 0.7697   <--- I think JIT Compiler effect. Java different from Python.. 
// Vectorization ms : 0.6571
```

### Notes : 
- NumPy looks Python but its ***calculate engine*** C/C++. Python calls ***np.dot*** function, C++ calculates with parallel proccessings. 
- If you calculate dot product with loop in Python, Python doesn't call C++. So calculation so slowly.

> We should updating models' parameters ***simultaneously***. So we have vectorization advantages.
