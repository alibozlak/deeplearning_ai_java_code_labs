### Optional : Gradient Descent Algorithm

If a function has derivative, and <br>
- $ g'(x_0) \lt 0 \Rightarrow \ function\ is\ decreasing\ around\ x_0\ point $ 
- $ g'(x_0) \gt 0 \Rightarrow \ function\ is\ increasing\ around\ x_0\ point $ 

(For proof : [Relation between Derivative and Functions Increase or Decrease ](../../../Math/Calculus/SingleVariableSingleValueFunctions/DerivativeAndFunctionIncreaseOrDecrease.ipynb)) <br>



#### Our cost (J) function :

So far, our model had 2 parameters : ***weight*** and ***bias***. But I will decrease 1 parameter for describe gradient descent with single variable calculus. <br>
- If we choose ***bias*** : $ \ f(x) = b, \qquad $ Model fits to datas by horizontal lines.<br>
- If we choose ***weight*** : $ \ f(x) = w \cdot x, \qquad $ Model fits to datas by lines have slope and $ \ \begin{bmatrix} 0 \\ 0 \end{bmatrix} \ $ point.<br>

Choosing ***weight*** parameter make sense : (w and x represent weight and feature value) <br>
$ f(x) = w \cdot x $<br>

So, our cost (and loss) function has only single variable : ***w*** <br>

$ J(w) = \frac{1}{2m} \sum_{i=0}^{m} L^{(i)}(w) = \frac{1}{2m} \sum_{i=0}^{m} (w \cdot x^{(i)} - y^{(i)})^2 $<br>

$ L^{(i)}(w) = (w \cdot x^{(i)} - y^{(i)})^2 \ $ function ***has derivative for all $w$ points***. Because $L^{(i)}\ $ is polynomial function. <br>

$ \Rightarrow J(w) = \frac{1}{2} \cdot arithmaticMean(L^{(i)}(w)) \ $ function ***has derivative for all $w$ points***. <br>

And ***both*** are has ***only single minimum value*** and ***both*** are ***convex*** functions. 
(Proof for loss function : [notebook 021](./021_optional_math_loss_function_only_single_local_value.ipynb))
(Proof for cost function : [notebook 022](./022_optinal_math_cost_func_only_single_local_value.ipynb))<br><br>

By the way, ***If our models multiple parameters we go same steps***. Because : $ (For\ example\ if\ f(x) = wx + b \ be) $ <br>

$ \frac{\partial J(w,b)}{\partial w} = \frac{dJ(w)}{dw} \qquad  (b\ is\ constant\ here) \ $ Stands for : How much does ***w*** effect on ***J***s changing ? <br>

$ \frac{\partial J(w,b)}{\partial b} = \frac{dJ(b)}{db} \qquad  (w\ is\ constant\ here) \ $ What about ***b*** ? <br><br>


### ***How we will train model?***

We need a algorithm. Good news is, we know these right now: <br> 
- Our loss (L) and cost (J) functions have ***single minimum value***,
- Both have derivative for ***all points*** and 
- We have a info about ***relation between derivative sign and function is increasing or decreasing***. By the way, Loss and Cost functions also convex. So If $\ \frac{dJ}{dw} = 0 \Rightarrow w\ $ is ***a local minimum point***. <br>

Let's continue with $\ f(x) = wx\ $ model : <br>

```java
Initialize with random w value;
Calculate the cost J(w); //For compare to new w values J 
Calculate derivative dJ_dw(w);  //Not: Andrew Ng is using "dw" notation

if (dJ_dw(w) == 0){
    We catched minimum value. Stop.
} 
else if (dJ_dw(w) < 0){
    J is decreasing around w point. If we increase the w, J will descrease. 
    BUT, how will we catch correct w's neighbor? (We have a step size problem)
    
    J will have different secure neighbour length by different w. So we should choose a step length dependents with w.
    Can we choose stepSize = dJ_dw(w) ?
    Make sense. But dJ_dw(w) < 0. So
       new_w = w + dJ_dw(); --> J(new_w) > J(w) !! This is a problem.
    
    We should choose stepSize = - dJ_dw(w).
        new_w = w - dJ_dw(w);   //<---- Updating of Model's w parameter.. 
    //Note: Not enough.. (Explain below.) But we advanced too much than our initial case.
}
else {
    // dJ_dw(w) > 0
    new_w = w + dJ_dw(w);  --> J(new_w) > J(w) !!  //Because J is increasing. //Same problem above..

    //Gradint Descent Algorithm's KEY :
    //*****************************************************************************************************/
    //********** Generalization : If we increase a point with its derivative, Function ALWAYS grow.. ******/
    //*****************************************************************************************************/

    We should choose stepSize = - dJ_dw(w) again.
        new_w = w - dJ_dw(w);
}

Loop the latest step.

```

***Short summary of Algorithm :*** 
```java
Initialize with random w value;
Loop {
    if (dJ_dw(w) == 0)
        break; 
    new_w = w - dJ_dw(w);

    w = new_w;  //<---- I wrote for Simultaneously update. We don't need this assign right now. But we will need on model has multi-parameter. So much important stuation.
}
```

#### Why did I say "Not enough" about ***new_w = w - dJ_dw(w) updating***?

````python
import deeplearning_ai.machine_learning_specialization.course1.week1.*;
import deeplearning_ai.math.calculus.single_variable_single_value.*;

import java.util.function.Function;
````

````python
Function<Double,Double> J = w -> Math.pow(w - 4, 2);   //--->     J(w) = (w - 4)^2 
Function<Double,Double> dJ_dw = w -> 2 * (w - 4);      //---> dJ_dw(w) = 2 * (w - 4)

double w = 6;
void applyGradientDescent(int loopSize, Function<Double,Double> updateParameter) {
    for (int i = 0; i < loopSize; i++) {
        double cost = J.apply(w);
        System.out.println("w = " + w + ", J(" + w + ") = " + cost);
        
        double derivative = dJ_dw.apply(w);
        System.out.println("dJ_dw(" + w + ") = " + derivative + "\n");
        double new_w = updateParameter.apply(w);
    
        w = new_w;   //<-- For adapting Simultaneously update.
    }
}

Function<Double,Double> wrongWay = w -> w - dJ_dw.apply(w);

applyGradientDescent(4, wrongWay);    //Result : J didn't decrease. w entried loop : 6 -> 2, next step 2 -> 6. J(w) was 4 each step. 
````

For solution : We will add a ***learning rate*** in algorithm. <br>

We will choose $\ learningRate \in (0,1)\ : $<br>
```java
    new_w = w - lerningRate * dJ_dw(w);
```


````python
double learningRate = 0.1;
Function<Double,Double> correctWay = w -> w - learningRate * dJ_dw.apply(w);

applyGradientDescent(6, correctWay);   

// Logs : 
// w = 6.0, J(6.0) = 4.0
// w = 5.6, J(5.6) = 2.5599999999999987
// w = 5.279999999999999, J(5.279999999999999) = 1.6383999999999983
// w = 5.023999999999999, J(5.023999999999999) = 1.0485759999999982
// w = 4.8191999999999995, J(4.8191999999999995) = 0.6710886399999991
// w = 4.65536, J(4.65536) = 0.4294967295999999
// w = 4.524288, J(4.524288) = 0.2748779069440003
// w = 4.4194304, J(4.4194304) = 0.17592186044416036
// w = 4.33554432, J(4.33554432) = 0.11258999068426263
// w = 4.268435456000001, J(4.268435456000001) = 0.07205759403792827
// w = 4.2147483648, J(4.2147483648) = 0.04611686018427394
// w = 4.17179869184, J(4.17179869184) = 0.029514790517935385
// w = 4.137438953472, J(4.137438953472) = 0.018889465931478694
// w = 4.1099511627776, J(4.1099511627776) = 0.012089258196146287
// w = 4.08796093022208, J(4.08796093022208) = 0.007737125245533561
// w = 4.070368744177664, J(4.070368744177664) = 0.004951760157141479
// w = 4.056294995342131, J(4.056294995342131) = 0.0031691265005705264
// w = 4.045035996273705, J(4.045035996273705) = 0.002028240960365153
// w = 4.036028797018964, J(4.036028797018964) = 0.0012980742146336851

// It works :)
// Let's celebrate this 
````
