### Cost Function (J) has only single local minimum value : 

Loss function (L) for ***single sample***. If we have ***m samples*** : 

$ i \in \set{1, 2, ...., m} $ <br> 

Our model : <br>
$ f \colon \mathbb{R} \to \mathbb{R}, \ f(x^{(i)}) = w \cdot x^{(i)} + b $ <br> 

Our loss function : <br>
$ L \colon \mathbb{R}^{2} \to \mathbb{R}^{\ge 0}, \ L^{(i)}(\begin{bmatrix} w \\ b \end{bmatrix}) = (f(x^{(i)}) - y^{(i)})^{2} $

Our cost function: 
$ J \colon \mathbb{R}^{2} \to \mathbb{R}^{\ge 0} $ <br>

$$J(\begin{bmatrix} w \\ b \end{bmatrix}) = \frac{1}{2m}\sum_{i=1}^{m}L^{(i)}(\begin{bmatrix} w \\ b \end{bmatrix})$$

$ We \ choose \ \textbf{arbitrary} \begin{bmatrix} w_1 \\ b_1 \end{bmatrix}, \begin{bmatrix} w_2 \\ b_2 \end{bmatrix} \in \mathbb{R}^{2} \ and \ \textbf{arbitrary} \ \lambda \in [0,1] \ .$ <br>

$L^{(i)}$ is the function $L$, $x^{(i)} = x$ and $y^{(i)} = y$ in [notebook 021](021_optional_math_loss_function_only_single_local_value.ipynb). In that notebook we showed: <br>

$ L^{(i)}(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ \le \ \lambda L^{(i)}(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) L^{(i)}(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) \ \ $ (For <b>all i</b>)<br> 

$$\sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ \ \le \ \ \lambda \sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1-\lambda) \sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix})$$

m > 0 : <br>

$$\frac{1}{2m}\sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ \ \le \ \ \lambda\frac{1}{2m}\sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1-\lambda)\frac{1}{2m}\sum_{i=1}^{m} L^{(i)}(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}$$

$$J(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ \ \le \ \ \lambda J(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1-\lambda) J(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix})$$

<br><br>
Cost function J is ***convex***. It has only single local minimum ***value***.
