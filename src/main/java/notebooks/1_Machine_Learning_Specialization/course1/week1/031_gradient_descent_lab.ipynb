### Gradient Descent Lab 

Our model $\ f(x) = w \cdot x + b \ $ <br>


```python
import deeplearning_ai.machine_learning_specialization.course1.week1.*;

import java.util.function.Function;
```

```python
//Our dataset : 
double[] sizeOfHouses = {1.0, 2.0, 1.5, 1.2, 0.8};  //Unit : 1000 square feet
double[] prices = {300., 500., 350., 300., 310.};   //Unit : 1000s of dollars

double weight = 100.;
double bias = 200.;
Function<Double,Double> f = area -> weight * area + bias;   //<--- Our model

double J() {
    int sampleCount = sizeOfHouses.length;
    double cost = 0.;
    for (int i = 0; i < sampleCount; i++){
        double modelPrediction = f.apply(sizeOfHouses[i]);
        double realPrice = prices[i];
        cost += Math.pow(modelPrediction - realPrice, 2);
    }
    return cost / (2*sampleCount);
}

double j = J();
System.out.println("f(x) = " + weight + " * x + " + bias + " model's cost = " + j);  
//f(x) = 100.0 * x + 200.0 model's cost = 1130.0
```

#### $\frac{\partial J(w,b)}{\partial w}\ $ and $\frac{\partial J(w,b)}{\partial b}\ $ Computes : 
$ m = sample\ size\ $<br>

$ L^{(i)}(w, b) = (f(x^{(i)}) - y^{(i)})^2 = (wx^{(i)} + b - y^{(i)})^2 $<br>

Let $\ z^{(i)}(w,b) = wx^{(i)} + b - y^{(i)}\ $ say, $\ L^{(i)}(w,b) = (z^{(i)}(w,b))^2 $<br><br>

- $\frac{\partial J(w,b)}{\partial w}\ $ Part : <br>

$ \frac{\partial L^{(i)}(w,b)}{\partial w}|_{(w_0,b_0)} = \frac{dL^{(i)}(w)}{dw}|_{w_0} = \frac{dL^{(i)}(w)}{dz^{(i)}}|_{z^{(i)}(w_0)} \cdot \frac{dz^{(i)}(w)}{dw}|_{w_0} $<br>

$ = 2\cdot z^{(i)}(w) \cdot x^{(i)} = 2\cdot (wx^{(i)} + b - y^{(i)}) \cdot x^{(i)} = 2\cdot (f(x^{(i)}) - y^{(i)}) \cdot x^{(i)} $<br>

$ J(w,b) = \frac{1}{2m}\cdot \sum_{i = 1}^{m} L^{(i)}(w, b) $<br>

$\frac{\partial J(w,b)}{\partial w} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} \frac{\partial L^{(i)}(w,b)}{\partial w} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} 2\cdot (f(x^{(i)}) - y^{(i)}) \cdot x^{(i)} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(x^{(i)}) - y^{(i)}) \cdot x^{(i)} $ <br> 

$$
\frac{\partial J(w,b)}{\partial w} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(x^{(i)}) - y^{(i)}) \cdot x^{(i)}
$$
<br>

- $\frac{\partial J(w,b)}{\partial b}\ $ Part : 

Same steps above. 

$ \frac{\partial L^{(i)}(w,b)}{\partial b} = \frac{dL^{(i)}(b)}{db} = \ chain\ rule\ = 2\cdot (f(x^{(i)}) - y^{(i)}) $<br>

$\frac{\partial J(w,b)}{\partial b} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} \frac{\partial L^{(i)}(w,b)}{\partial b} = \frac{1}{2m}\cdot \sum_{i = 1}^{m} 2\cdot (f(x^{(i)}) - y^{(i)}) = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(x^{(i)}) - y^{(i)}) $ <br>

$$
\frac{\partial J(w,b)}{\partial b} = \frac{1}{m}\cdot \sum_{i = 1}^{m} (f(x^{(i)}) - y^{(i)})
$$
<br>

#### Model's training : 

$ new_w = w - learningRate\cdot \frac{\partial J(w,b)}{\partial w} $<br>

$ new_b = b - learningRate\cdot \frac{\partial J(w,b)}{\partial b} $<br>

```python
double dJ_dw() {
    int m = sizeOfHouses.length;
    double result = 0;
    for (int i = 0; i < m; i++){
        double area_i = sizeOfHouses[i];
        double realPrice_i = prices[i];
        result += area_i * (f.apply(area_i) - realPrice_i);
    }
    return result / m;
}
```

```python
double dJ_db() {
    int m = sizeOfHouses.length;
    double result = 0;
    for (int i = 0; i < m; i++){
        double area_i = sizeOfHouses[i];
        double realPrice_i = prices[i];
        result += f.apply(area_i) - realPrice_i;
    }
    return result / m;
}
```

```python
void applyGradientDescent(int loopSize, double learningRate) {
    System.out.println("w = " + weight + ", b = " + bias + ", J(w,b) = " + J());

    for (int i = 0; i < loopSize; i++) {
        double new_w = weight - learningRate * dJ_dw();
        double new_b = bias   - learningRate * dJ_db();

        weight = new_w;
        bias = new_b;

        System.out.println("w = " + ((int)weight) + ", b = " + ((int)bias) + ", J(w,b) = " + J());
    }
}

applyGradientDescent(4, 0.1);

// Logs : 
// w = 100.0, b = 200.0, J(w,b) = 1130.0            Initial model : f(x) = 100*x + 200
// w = 104, b = 202, J(w,b) = 950.3880000000003     f(x) = 104*x + 202
// w = 106, b = 203, J(w,b) = 855.9204114260804
// w = 109, b = 204, J(w,b) = 805.5703003245354
// w = 110, b = 205, J(w,b) = 778.0880677954353     f(x) = 110*x + 205   //4. iteration
// .. 
// w = 120, b = 199, J(w,b) = 699.8574495296004     f(x) = 120*x + 199   //24. iteration
// .. 
// w = 164, b = 137, J(w,b) = 518.7962834226346     f(x) = 164*x + 137   //1_000. iteration
// .. 
// w = 164, b = 137, J(w,b) = 518.7954545454547     f(x) = 164*x + 137   //4_800.  <--- Enough count
// ..
// -------------- Same Cost ----------------- 
// ..
// w = 164, b = 137, J(w,b) = 518.7954545454547     f(x) = 164*x + 137   //100_000. iteration 


// Total loop count: 100_000  (Enough : 4_800) (Training was slow, because we didn't scale)
// Last parameters : w = 164.77272727272555, b = 137.79545454545692   <--- double precision
// Our model       : f(x) = w*x + b  
```
