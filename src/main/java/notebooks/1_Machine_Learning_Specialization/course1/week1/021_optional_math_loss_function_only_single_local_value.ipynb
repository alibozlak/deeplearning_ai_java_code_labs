### Loss Function (L) has only single local minimum value : 

Our model : <br>
$ f \colon \mathbb{R} \to \mathbb{R}, f(x) = w \cdot x + b $ was. <br> 

Our loss function : <br>
$ L \colon \mathbb{R}^{2} \to \mathbb{R}^{\ge 0}, L(\begin{bmatrix} w \\ b \end{bmatrix}) = (f(x) - y)^{2} $ was.

Question : ***Does L have only single local minimum value?*** <br>
This question important. Because we will train (learning from L or J) to model ***with derivative***. <br>

1. If L has not minimum value : We can't decrease error. 
2. If L has greater than 1 local minimum value : Model (f) can optimize by ***non-global minimum value*** itself

> If we can L function is convex, we show L has only one local minimum **value**.

We should show <br>

$ All \begin{bmatrix} w_1 \\ b_1 \end{bmatrix}, \begin{bmatrix} w_2 \\ b_2 \end{bmatrix} \in \mathbb{R}^{2} \ and \ all \ \lambda \in [0,1] $ should valid that rule :  <br><br>
$ L(\begin{bmatrix} \lambda \cdot w_1 + (1 - \lambda) \cdot w_2 \\ \lambda \cdot b_1 + (1 - \lambda) \cdot b_2\end{bmatrix}) \le \lambda \cdot L(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) \cdot L(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) $ <br><br>

$ We \ choose \ \textbf{arbitary} \begin{bmatrix} w_1 \\ b_1 \end{bmatrix}, \begin{bmatrix} w_2 \\ b_2 \end{bmatrix} \in \mathbb{R}^{2} \ and \ \textbf{arbitary} \ \lambda \in [0,1] $ <br>

$ Left \ side \ of \ inequality \ : $ <br>

$ L(\begin{bmatrix} \lambda \cdot w_1 + (1 - \lambda) \cdot w_2 \\ \lambda \cdot b_1 + (1 - \lambda) \cdot b_2\end{bmatrix}) = ((\lambda \cdot w_1 + (1 - \lambda) \cdot w_2) \cdot x + \lambda \cdot b_1 + (1 - \lambda) \cdot b_2 - y)^{2} $ <br>

$ (We \ can \ write \ -y = -\lambda y -(1-\lambda)y) $<br>

$ = ((\lambda w_1 \cdot x + (1-\lambda)w_2 \cdot x) + \lambda b_1  + (1-\lambda)b_2 -\lambda y -(1-\lambda)y)^{2} $ <br>

$ = (\lambda(w_1 x + b_1 - y) + (1-\lambda)\cdot(w_2 x + b_2 - y))^{2} $ <br>

$ A := f(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) = w_1\cdot x + b_1 \ and \ B := f(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) = w_2\cdot x + b_2 $ <br>

$ = (\lambda (A - y) + (1-\lambda)(B - y))^{2} $ <br>

$ = \lambda^{2}(A - y)^{2} + (1-\lambda)^{2}(B - y)^{2} + 2\lambda(1-\lambda)(A - y)(B - y) $ <br>

$ L(\begin{bmatrix} \lambda \cdot w_1 + (1 - \lambda) \cdot w_2 \\ \lambda \cdot b_1 + (1 - \lambda) \cdot b_2\end{bmatrix}) = \lambda^{2}(A - y)^{2} + (1-\lambda)^{2}(B - y)^{2} + 2\lambda(1-\lambda)(A - y)(B - y) $ <br><br>

$ Right \ side \ of \ inequality :  $<br>

$ \lambda \cdot L(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) \cdot L(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) = \lambda \cdot (A - y)^{2} + (1-\lambda) \cdot (B - y)^{2} $ <br><br>

$ Right \ side \ - \ left \ side \ : $ <br>

$ \lambda L(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) L(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) - L(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) = \lambda (A - y)^{2} + (1-\lambda) (B - y)^{2} - (\lambda^{2}(A - y)^{2} + (1-\lambda)^{2}(B - y)^{2} + 2\lambda(1-\lambda)(A - y)(B - y)) $ <br>

$ = \lambda (A - y)^{2} - \lambda^{2}(A - y)^{2} + (1-\lambda) (B - y)^{2} - (1-\lambda)^{2}(B - y)^{2} - 2\lambda(1-\lambda)(A - y)(B - y) $ <br>

$ = \lambda(A - y)^{2}(1 - \lambda) + (1-\lambda)(B - y)^{2}(1 - (1-\lambda)) - 2\lambda(1-\lambda)(A - y)(B - y) $ <br>

$ = \lambda(A - y)^{2}(1 - \lambda) + (1-\lambda)(B - y)^{2}\lambda - 2\lambda(1-\lambda)(A - y)(B - y) $ <br>

$ = \lambda(1-\lambda)((A-y)^2 + (B-y)^{2} - 2(A-y)(B-y)) $ <br>

$ = \lambda(1-\lambda)(A-y - (B-y))^{2} $ <br>

$ = \lambda(1-\lambda)(A-B)^{2} $ <br>

$ (\lambda \in [0,1] \ . \ So \ (1-\lambda) \in [0,1] \ . \ And \ (A-B)^{2} \ge 0 \ .) $ <br>

$ \lambda L(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) L(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) - L(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ge 0 $ <br>

$ L(\begin{bmatrix} \lambda w_1 + (1 - \lambda) w_2 \\ \lambda b_1 + (1 - \lambda) b_2\end{bmatrix}) \ \le \ \lambda L(\begin{bmatrix} w_1 \\ b_1 \end{bmatrix}) + (1 - \lambda) L(\begin{bmatrix} w_2 \\ b_2 \end{bmatrix}) $ <br>

$ \begin{bmatrix} w_1 \\ b_1 \end{bmatrix}, \begin{bmatrix} w_2 \\ b_2 \end{bmatrix} \in \mathbb{R}^{2} \ and \ \lambda \in [0,1] \ \textbf{arbitary} \ . \ So \ L \ function \ is \ \textbf{convex} \ .$ <br>


$ So \ loss \ function \ L \ has \ only \ single \ local \ minimum \ \textbf{value}. $
