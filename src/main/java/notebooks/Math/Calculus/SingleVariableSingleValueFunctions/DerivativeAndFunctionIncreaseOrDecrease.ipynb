### Relation between Derivative and Functions Increase or Decrease 

$ Let \ J \colon \mathbb{R} \to \mathbb{R} \ and \ J, \ has \ derivative \ a \ w_0\in\mathbb{R} \ point.   $<br>

If $ \ \frac{dJ(w)}{dw}|_{w = w_0} \lt 0 \Rightarrow J, \ decreasing \ around \ w_0 \ point. $  <br>

If $ \ \frac{dJ(w)}{dw}|_{w = w_0} \gt 0 \Rightarrow J, \ increasing \ around \ w_0 \ point. $  <br>

```python
import deeplearning_ai.math.calculus.single_variable_single_value.*;

import static java.lang.System.out;
```

1) Let $ \ \frac{dJ(w)}{dw}|_{w = w_0} \lt 0 \ be. $ <br>

From derivatives ***limit definition*** : $ \ lim_{\Delta w \to 0} \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w} \lt 0 $<br>

$ H(\Delta w) := \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w}, \qquad lim_{\Delta w \to 0} H(\Delta w) \lt 0  $<br>

$ There \ is \ exist \ a \ \varepsilon\gt 0, \ all \ \Delta w \in (-\varepsilon,\varepsilon)-\set{0} \qquad H(\Delta w) \lt 0 $ <br>

$ \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w} \lt 0 $<br>

- If $ \ \Delta w \gt 0, \qquad J(w_0 + \Delta w) - J(w_0) \lt 0 $<br>

$ J(w_0 + \Delta w) \lt J(w_0) \qquad (J \ is \ decreasing \ in \ [w_0, w_0 + \varepsilon) ) $ <br>

- If $ \ \Delta w \lt 0, \qquad J(w_0 + \Delta w) - J(w_0) \gt 0 $

$ J(w_0 + \Delta w) \gt J(w_0) \qquad (J \ is \ decreasing \ in \ (w_0 - \varepsilon, w_0]) $ <br>

$ J \ is \ decreasing \ in \ (w_0 - \varepsilon, w_0 + \varepsilon). $ <br><br>

```python
float sizeOfHouseSample1 = 1.f;
float realPriceSample1 = 300.f;
J cost = J.getFunction(sizeOfHouseSample1, realPriceSample1);

float w = 275.f;
out.println("dJ_dw(" + w + ") = " + cost.calculateDerivative(w));   //dJ_dw(275.0) = -50.0
```

```python
//Support method 
void printFunctionValuesAround(J func, float w, int times) {
    float epsilon = 0.01f;
    float step = epsilon / 10;
    float initialPoint = w - epsilon;
    for (int t = 0; t < times; t++){
        float smallPointsValue = func.apply(initialPoint);
        float greatPointsValue = func.apply(initialPoint + step);
        out.println("J(" + initialPoint + ") = " + smallPointsValue + ", J(" + (initialPoint + step) + ") = " + greatPointsValue);
        
        step *= 2;
        initialPoint += step;
    }
}
```

```python
printFunctionValuesAround(cost, w, 7);
```

2) Let $ \ \frac{dJ(w)}{dw}|_{w = w_0} \gt 0 \ be. \ $ Derivatives limit definition: <br>

$ \ lim_{\Delta w \to 0} \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w} \gt 0 $<br>

$ H(\Delta w) := \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w}, \qquad lim_{\Delta w \to 0} H(\Delta w) \gt 0  $<br>

$ There \ is \ exist \ a \ \varepsilon\gt 0, \ all \ \Delta w \in (-\varepsilon,\varepsilon)-\set{0} \qquad H(\Delta w) \gt 0 $ <br>

$ \frac{J(w_0 + \Delta w) - J(w_0)}{\Delta w} \gt 0 $<br>

- If $ \ \Delta w \gt 0, \qquad J(w_0 + \Delta w) - J(w_0) \gt 0 $<br>

$ J(w_0 + \Delta w) \gt J(w_0) \qquad (J \ is \ increasing \ in \ [w_0, w_0 + \varepsilon)) $ <br>

- If $ \ \Delta w \lt 0, \qquad J(w_0 + \Delta w) - J(w_0) \lt 0 $

$ J(w_0 + \Delta w) \lt J(w_0) \qquad (J \ is \ increasing \ in \ (w_0 - \varepsilon, w_0] $ <br>

$ J\ is \ increasing \ in \ (w_0 - \varepsilon, w_0 + \varepsilon). $ <br><br>

```python
w = 320.f;
out.println("dJ_dw(" + w + ") = " + cost.calculateDerivative(w));   //dJ_dw(320.0) = 40.0

printFunctionValuesAround(cost, w, 7);
```
