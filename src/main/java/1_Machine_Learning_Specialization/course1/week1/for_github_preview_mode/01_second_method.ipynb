## Second method For Predict Function :

```java
import org.nd4j.linalg.api.ndarray.INDArray;
import org.nd4j.linalg.factory.Nd4j;

import java.util.function.Function;

import deeplearning_ai.machine_learning_specialization.course1.week1.*;
```

```java
INDArray features = Nd4j.create(new float[] {1.f, 2.f});    // [1, 2]      (1000 sqft)
INDArray targets = Nd4j.create(new float[] {300.f, 500.f}); // [300, 500]  (1000s of dollars) 

long sampleSize = ValidateDataSet.validateFeatureArraySizeAndTargetArraySize(features, targets);

float weight = 100.f;
float bias = 100.f;

Function<INDArray, INDArray> singleVariableLinearFunction = new SingleVariablePredictFunction(weight, bias);
INDArray predicts = singleVariableLinearFunction.apply(features);

System.out.println("Predicts = " + predicts + " (1000s of dollars)");    //Predicts = [  200.0000,  300.0000] (1000s of dollars)

System.out.println("Real targets = " + targets + " (1000s of dollars)"); //Real targets = [  300.0000,  500.0000] (1000s of dollars)
```

We have ***just 2 samples***. So actually we don't need machine learning tecniquies :) <br>
2 points describe a line. We can find $w$ and $b$ with ***analytic geometry*** or ***basic algebra*** : <br>
$point1 = \begin{bmatrix} 1 \\ 300 \end{bmatrix} $ and $point2 = \begin{bmatrix} 2 \\ 500 \end{bmatrix} $ <br>

`Analytic Geometry :` <br>

$slope = w = \frac{point2_{target} - point1_{target}}{point2_{feature} - point1_{feature}} $ 

```java
weight = (targets.getFloat(1) - targets.getFloat(0)) / (features.getFloat(1) - features.getFloat(0));
System.out.println("My w = " + weight);   //My w = 200.0
```

$(x^{(1)},y^{(1)}) = (1, 300) $ <br>

So $f(1) = 300 $. <br> 

$f(1) = w \cdot 1 + b  = 200 \cdot 1 + b = 300 $ <br>

$b = 100 $

```java
bias = 100.0f;

singleVariableLinearFunction = new SingleVariablePredictFunction(weight, bias);
```

$(x^{(2)},y^{(2)}) = (2, 500) $ <br>

So $f(2) $ should be $500$ :  

```java
INDArray ourPredicts = singleVariableLinearFunction.apply(features);
float ourSecondSamplePredict = ourPredicts.getFloat(1);

System.out.println("Our predict = " + ourSecondSamplePredict + ", Real = " + 500);      //Our predict = 500.0, Real = 500
```
