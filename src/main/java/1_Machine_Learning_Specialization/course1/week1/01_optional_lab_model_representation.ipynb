## Notation : 

- $a$ ---> scalar value 
- $\vec{a}$ or $\mathbf{a} $ (bold) ---> vector value <br>
Regression : 
- $\vec{x}$ ---> Training Example feature values (in this lab - Size or Area (1000 square feet)) ---> `float[] samplesFeatureValues;` <br>
- $\vec{y}$ ---> Training Example target values (in this lab Price (1000s of dollars)) ---> `float[] samplesTargetValues;` <br>
- $m$ ---> Number of training examples (sample size) ---> `int sampleCount = samplesFeatureValues.lenght;` <br>
- $x^{(i)}$ ---> $i^{th}$ sample feature value ---> `float iThSampleFeatureValue = samplesFeatureValues[i];` <br>
- $y^{(i)}$ ---> $i^{th}$ sample target value ---> `float iThSampleTargetValue = samplesTargetValues[i];` <br>
- $f$ ---> Prediction linear ***function***, also known ***hypothesis***  ($f(x) = w\cdot x + b$) <br> ----> 
`java.util.function.Function<Float,Float> singleVariablePredictionFunction = x -> w * x + b;` <br>
- $w$ ---> $f$'s ***weight*** ---> It is a ***scalar*** value
- $b$ ---> $f$'s ***bias*** ---> It is a ***scalar*** value
- $f_{w,b}(x^{(i)}) = f(x^{(i)}) = w \cdot x^{(i)} + b$ ---> $i^{th}$ sample target predict <br> 
---> `float predictForIThSample = singleVariablePredictionFunction.apply(samplesFeatureValues[i]);` 

## Tools : 
- ND4J : NumPy Like Java dependency  ---> `import org.nd4j.*;`


```java
import org.nd4j.linalg.api.ndarray.INDArray;
import org.nd4j.linalg.factory.Nd4j;
```

### We have 2 samples : (1000 sqft, 300_000 dollars) and (2000 sqft, 500_000 dollars). 

```java
INDArray features = Nd4j.create(new float[] {1.f, 2.f});    // [1, 2]      (1000 sqft)
INDArray targets = Nd4j.create(new float[] {300.f, 500.f}); // [300, 500]  (1000s of dollars) 

long sampleSize = features.length();
if (sampleSize != targets.length())
    throw new RuntimeException("Features size should be equal to Targets size !!");

System.out.println("Sample size m = " + sampleSize);    //Sample size m = 2
```

```java
float weight = 100.f;
float bias = 100.f;

java.util.function.Function<Float,Float> singleVariablePredictFunction = featureValue -> weight * featureValue + bias;
float firstSampleFeature = features.getFloat(0);
float predictTargetFirstSample = singleVariablePredictFunction.apply(firstSampleFeature);

System.out.println("First sample's feature = " + firstSampleFeature + "* 1000 sqft, Function predict = " + predictTargetFirstSample + 
    "*1000 dollars, Real = " + targets.getFloat(0) + "*1000 dollars"
);
//First sample's feature = 1.0* 1000 sqft, Function predict = 200.0*1000 dollars, Real = 300.0*1000 dollars
```

```java
java.util.function.Function<INDArray,INDArray> singleVariablePredictFunctionForArray = featureArray -> {
    int m = (int) featureArray.length();
    float[] predictArray = new float[m];
    for (int i = 0; i < m; i++)
        predictArray[i] = singleVariablePredictFunction.apply(featureArray.getFloat(i));
    return Nd4j.create(predictArray);
};
```

```java
INDArray predictArray = singleVariablePredictFunctionForArray.apply(features);
System.out.println("predictArray = " + predictArray);      //predictArray = [  200.0000,  300.0000]
System.out.println("targets = " + targets);                //targets = [  300.0000,  500.0000]
```
